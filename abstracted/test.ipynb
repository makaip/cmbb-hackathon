```python
import pandas as pd

path = "../23Q_counts_data ex.xlsx"
df = pd.read_excel(path)
```

```python
df = df[df['23Q'] > 1]
df.reset_index(drop=True, inplace=True)
df
```

```python
from readcount import ReadCountProcessor

processor = ReadCountProcessor(df, gene_limit=100)
normalized_df = processor.normalize_read_counts(max_workers=100)
normalized_df
```

```python
from geneconverter import convert_dataframe_parallel

df_with_ncbi = convert_dataframe_parallel(normalized_df, row_limit=200, max_workers=100)
df_with_ncbi = df_with_ncbi[pd.to_numeric(df_with_ncbi['NCBI_ID'], errors='coerce').notnull()]
df_with_ncbi
```

```python
df_with_ncbi_sorted = df_with_ncbi.sort_values(by='Normalized_Read_Counts', ascending=False)
df_with_ncbi_sorted = df_with_ncbi_sorted.iloc[:20]
df_with_ncbi_sorted
```

```python
from geneprocessor import GeneDataProcessor

df_with_ncbi_sorted = GeneDataProcessor.sort_by_read_counts(df_with_ncbi)
df_with_gene_columns = GeneDataProcessor.add_gene_columns(df_with_ncbi_sorted)
df_final = GeneDataProcessor.enrich_gene_data(df_with_gene_columns)
```
